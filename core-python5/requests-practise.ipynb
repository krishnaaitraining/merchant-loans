```python
import requests
resp = requests.get("https://api.github.com/repos/psf/requests")
print(resp.status_code) # 200
print(resp.headers['Content-Type']) # application/json; charset=utf-8
repo = resp.json() # Automatic JSON decode
print(repo['stargazers_count'])
```

```python
params = {"q": "python", "sort": "stars", "per_page": 5}
resp = requests.get("https://api.github.com/search/repositories", params=params)
print(resp.url) # Encoded URL with ?q=python&sort=stars&per_page=5
print(resp.status_code) # 200
print(resp.headers['Content-Type']) # application/json; charset=utf-8
repos = resp.json() # Automatic JSON decode
print(repos['total_count']) # Total number of repositories
print(repos['items']) # List of repositories
print(repos['items'][0]['full_name']) # Full name of the first repository
print(repos['items'][0]['stargazers_count']) # Number of stars for the first repository
print(repos['items'][0]['html_url']) # URL of the first repository
print(repos['items'][0]['description']) # Description of the first repository
print(repos['items'][0]['language']) # Language of the first repository
```
